# 02 — EDA & Business Insights

This notebook converts cleaned records into management-oriented metrics covering donors, beneficiaries, volunteers and program efficiency.

```python
import pandas as pd
import plotly.express as px

d = pd.read_csv("../data/donations.csv")
b = pd.read_csv("../data/beneficiaries.csv")
v = pd.read_csv("../data/volunteers.csv")
p = pd.read_csv("../data/programs.csv")
```

```python
donor_channel = d.groupby("channel", as_index=False).agg(
    donations=("donation_id","count"),
    funds=("amount_inr","sum"),
    average_donation=("amount_inr","mean")
)
display(donor_channel.round(2))
px.bar(donor_channel, x="channel", y="funds", title="Funds by Donation Channel")
```

```python
impact = b.groupby("program", as_index=False).agg(
    beneficiaries=("beneficiary_id","count"),
    attendance=("attendance_rate","mean"),
    outcome=("outcome_score","mean")
)
display(impact.round(2))
px.scatter(impact, x="attendance", y="outcome", size="beneficiaries",
           hover_name="program", title="Attendance vs Outcome")
```

```python
volunteer = v.groupby("program", as_index=False).agg(
    volunteers=("volunteer_id","count"),
    total_hours=("hours_contributed","sum")
)
display(volunteer.round(2))
```
