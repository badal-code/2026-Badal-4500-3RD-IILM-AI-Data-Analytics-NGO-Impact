# 01 — NGO Impact Analytics: Exploratory Data Analysis

This notebook performs initial data loading, validation, cleaning and exploratory analysis on the synthetic demonstration datasets.

```python
import pandas as pd
import plotly.express as px

donations = pd.read_csv("../data/donations.csv")
beneficiaries = pd.read_csv("../data/beneficiaries.csv")
volunteers = pd.read_csv("../data/volunteers.csv")
programs = pd.read_csv("../data/programs.csv")

display(donations.head())
display(beneficiaries.head())
display(volunteers.head())
display(programs)
```

```python
print("Donations:", donations.shape)
print("Beneficiaries:", beneficiaries.shape)
print("Volunteers:", volunteers.shape)
print("Missing donation values:")
print(donations.isna().sum())
```

```python
donations["donation_date"] = pd.to_datetime(donations["donation_date"])
donations["month"] = donations["donation_date"].dt.to_period("M").astype(str)

monthly = donations.groupby("month", as_index=False)["amount_inr"].sum()
px.line(monthly, x="month", y="amount_inr", markers=True, title="Monthly Donation Trend")
```

```python
program_funds = donations.groupby("program", as_index=False)["amount_inr"].sum()
px.bar(program_funds, x="program", y="amount_inr", title="Donation Amount by Program")
```
