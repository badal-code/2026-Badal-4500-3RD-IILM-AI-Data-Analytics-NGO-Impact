# 03 — Donor Segmentation with K-Means

This notebook demonstrates unsupervised machine learning using aggregated donor-type features. The resulting clusters are exploratory segments, not labels about real individuals.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import plotly.express as px

d = pd.read_csv("../data/donations.csv")
features = d.groupby("donor_type", as_index=False).agg(
    total_amount=("amount_inr","sum"),
    avg_amount=("amount_inr","mean"),
    donation_count=("donation_id","count"),
    repeat_rate=("repeat_donor","mean")
)
features
```

```python
X = features[["total_amount","avg_amount","donation_count","repeat_rate"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

k = min(4, len(features))
model = KMeans(n_clusters=k, random_state=42, n_init=10)
features["segment"] = model.fit_predict(X_scaled)

display(features.round(3))
```

```python
px.scatter(
    features,
    x="avg_amount",
    y="total_amount",
    size="donation_count",
    color="segment",
    hover_name="donor_type",
    title="Exploratory Donor Segments"
)
```

### Note

K-Means is used here as a learning demonstration. With real donor-level data, segmentation should use an appropriate privacy-preserving feature set and be validated before operational use.
